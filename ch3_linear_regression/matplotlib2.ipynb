```python
# A line graph is great for showing a trend over a continuous sequence like sales over time.
# But sometimes you're comparing separate, distinct categories that don't really "flow" into
# each other, like sales per product, or population per country. For that, a bar chart
# communicates the comparison much more clearly than a line.

# bar charts are drawn with the plt.bar(x, y) function.
```

```python
# example
products = ["apples", "bananas", "cherries", "dates"]
units_sold = [40, 65, 30, 50]

plt.bar(products, units_sold)

plt.title("Units Sold by Product")
plt.xlabel("Product")
plt.ylabel("Units Sold")

plt.show()
```

```python
# example 2
# Bar chart for a small survey: favourite fruits among groups of people.
categories = ["Apple", "Banana", "Mango", "Orange"]
votes = [15, 22, 9, 18]

plt.bar(categories, votes, color="green")
plt.title("Votes by Category")
plt.xlabel("Fruit Category")
plt.ylabel("Votes")
plt.show()
```
