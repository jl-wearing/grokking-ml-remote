# Pandas Tutorial
## What is it?

Pandas is a library for working with tabular data, i.e. data organised into rows and columns.
If matplotlib is how you draw data, then pandas is how you load, organise, and inspect it before
you draw/model it.

Pandas works with **DataFrames**. A DataFrame is Python's version of a spreadsheet, i.e. rows
are individual records and columns are the different pieces of information about each record.

````python
data = {
    "name": ["Alice", "Bob", "Carol"],
    "age": [24, 30, 22],
    "city": ["Gaborone", "Maun", "Francistown"],
}

# creating a dataframe out of a dictionary.
df = pd.DataFrame(data)
print(df)
````

Each key becomes a column, and each value becomes that column's data.

````python
# example
books = {
    "title": ["spring start here", "grokking ml", "spring security in action",
              "big java"],
    "author": ["laurentiu spilca", "luis serrano", "laurentiu spilca",
               "cay horstmann"],
    "pages": [500, 600, 500, 900],
}

# creating a dataframe out of a dictionary.
df = pd.DataFrame(books)
print(df)
````

## Exploring a DataFrame
When you load real data, it's often hundreds or thousands of rows which is far too much
to read by just printing the whole thing. So before doing anything else, you always run
a few standard "get to know your data" commands.

### .head()
Shows the first 5 rows by default, or pass a number like .head(2) for a different amount.
This is like flipping open the first page of a book to see what you're dealing with, without
reading the whole thing.

### .shape
Returns the number of rows and columns as a tuple. Note that there are no parentheses.
.shape is an instance variable, not a function. It's like asking how big a spreadsheet is
without having to open it.

### .info()
Produces a summary of **column names**, **how many non-empty values each column has**, and the
**data type** of each column. This is your first check for **missing data**.

````python
# display the first few rows of the data.
print(df.head(), end="\n\n")

# display the number of rows and columns of the data.
print(df.shape, end="\n\n")

# display information about the columns
df.info()
````

### .describe()
Gives you a summary for numeric columns in one call. It gives you:
- count
- standard deviation (std)
- mean
- min
- max
- and others...

**.info()** is the table of contents (structure), while .describe() is the "at a glance" view.


````python
print(df.describe())
````

## Selecting Columns
This is where you learn to pull out specific pieces of a DataFrame instead of just looking
at the whole thing.

### Selecting One Column
Remember, a table is essentially just a 2d-matrix, so the notation is very similar to
accessing array elements. For example, **_df["title"]_** retrieves the column "title" from
the DataFrame.
This returns a **Series**, which is pandas' name for a single column of data.

### Selecting Multiple Columns
You do this by passing a list of column names, like **_df[["title", "author"]]_**.
This returns a full DataFrame (not a Series), just with fewer columns.

````python
# retrieving one column (series).
print(df["author"], end="\n\n")

# retrieving two columns (dataframe)
print(df[["title", "pages"]])
````

## Selecting Rows
There are 2 tools used to retrieve rows:
- **`.iloc[]`** and,
- **`.loc[]`**

**`.iloc[]`** selects by integer position, i.e. by row index.

**`.loc[]`** selects by label. It looks at the actual index value of the row, i.e. the
numbers on the left side of your DataFrame, or a custom label if you set one, not its position.

By default, the indexes happen to be the same, so both `.loc[]` and `.iloc[]` retrieve the same
thing. The difference only matters once you filter or sort a DataFrame (the index labels stay
attached to their original rows, while positions shift).

**Analogy**: Imagine a classroom of students seated in a row. `.iloc[2]` means "give me
whoever is sitting in the 3rd chair", i.e. it doesn't matter who that is, while `.loc[2]` means
"give me the student wearing the name-tag #2", i.e. even if they've sincemoved to a different
seat, you're finding them by their tag, not their current seat.

````python
print(df.iloc[0])
````

````python
print(df.loc[0])
````

### Selecting a Range of Rows
It is important to be aware of the differences between how `.iloc[]` and `.loc[]` behave
when retrieving a range of rows.
- `.iloc[]` retrieves from the starting index, upto but not including the end index.
- `.loc[]` retrieves all rows in the range (end is INCLUDED).

````python
print(df.iloc[0:2])
````

````python
print(df.loc[0:2])
````

### Selecting Rows and Columns Together


````python
print(df.loc[0:1, ["title", "pages"]])
````

````python
print(df.iloc[0:2, 0:2])
````

````python
# example - get the third row using .iloc
print(df.iloc[2], end="\n\n\n")

# - get rows 1 through 3 and the title and author columns.
print(df.loc[1:3, ["title", "author"]])
````

### Filtering Rows with Conditions (Boolean Indexing)
This is where pandas starts feeling like a real querying tool. For example, you can ask
"give me only the rows where some condition is true", e.g. "only books over 550 pages".

The resulting expression would look like `df["pages"] > 550`. This produces a Series of
`True/False` values, one per row.
The next step is to put that Series back into a DataFrame as `df[df["pages"] > 550]`.
Pandas only keeps the rows marked True.

You can also combine to form multiple conditions with **_&_** for AND, **_|_** for OR. Each
condition must be wrapped in its own parentheses (a common source of errors if you forget this)

````python
var = df[df["pages"] > 550]
print(var)

print('\n\n')
print(df)
````

````python
# forming larger boolean expressions.
var = df[(df['pages'] >= 500) & (df['author'] == 'laurentiu spilca')]
print(var)
````

````python
# example
long_books = df[df["pages"] > 500]
spilca_long_books = df[(df["pages"] == 500) & (df["author"] == "laurentiu spilca")]

print(long_books, spilca_long_books, sep="\n\n\n")
````

## Adding New Columns
You can create a new column just by assigning to a column name that doesn't exist yet.
Pandas will add it as a new column, computed from whatever expression you gave it (usually
derived from existing columns).

````python
# example
df["long_book"] = df["pages"] > 500
print(df)
````

````python
df["pages_per_hour"] = df["pages"] / 40 # vectorized operation
print(df)
````

## More Custom Transformations with `.apply()`
Sometimes the new column needs logic more complex than simple math. For that, you write a small
function and `.apply()` it to a column.

```
def size_label(pages):
    if pages >= 700:
        return "very long"
    return "short"
```
`df["size_category"] = df["pages"].apply(size_label)`

`.apply()` runs your function once per row's value in that column, and collects all the
results into a new column. It's the tool you reach for whenever the transformation is too
complex for a single line of math.

````python
# example.
df["is_spilca"] = df["author"] == "laurentiu spilca"
print(df)
````

````python
def categorize_book(number_of_pages: int):
    if number_of_pages < 600:
        return "short"
    return "long"


df["page_category"] = df["pages"].apply(categorize_book)
print(df)
````

## Splitting into Features and Target - The Bridge into scikit-learn
Every scikit-learn model expects data in exactly this shape.
In supervised machine learning, you're trying to predict one column, called the **target**
using other columns, called **features**.

````python
features = df[["pages", "is_spilca"]]   # needs to be a dataframe, even if only 1 column
target = df["page_category"]            # series is fine and expected.
````

Another way to build the feature set by dropping the target column instead of listing
every feature by name (useful when you have many columns and don't want to type them all out.)

axis=1 means drop a column, as opposed to axis=0, which would drop a row.

````python
features = df.drop("page_category", axis=1)
target = df["page_category"]

print(features, target, sep="\n\n\n")
````
