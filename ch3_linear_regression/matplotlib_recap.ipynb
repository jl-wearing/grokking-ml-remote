```python
x_coordinates = [1,2,3,4, 5]
y_coordinates = [100, 200, 300, 400, 500]

# plot a line graph.
# build the graph in memory.
plt.plot(x_coordinates, y_coordinates)

# show the graph.
plt.show()
```

```python
# Bar charts
fruits = ["Apples", "Bananas", "Cherries", "Dates", "Plums"]
units_sold = [300, 250, 185, 100, 200]

plt.bar(fruits, units_sold, color='green')
plt.title("Units Sold by Category")
plt.xlabel("Fruit Category")
plt.grid(True)
plt.ylabel("Units Sold")
plt.show()
```

```python
# Scatter Plots
study_hours = [1,2,3,4,5,6,7,8]
exam_scores = [50,55,65,60,70,75,85,90]
plt.scatter(study_hours, exam_scores, color='purple', s=100)
plt.title("Study Hours vs Exam Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Scores")
plt.show()
```

```python
# Multiple lines on one chart
# Often you want to compare 2 or more series directly, for example, this year's sales vs. last
# year's sales. You do this by simply building each graph in memory, i.e. calling plt.plot() or
# plt.scatter or bar before calling plt.show().

# There is one problem however => If you draw 2 lines, how does anyone know which line is
# which? That's what plt.legend() is for. It adds a little box showing which color/style
# maps to which label.

# How labeling works:
# You supply a label in the call to plot, then a plt.legend() call collects the labels into
# a legend box.
```

```python
# example
days = [1,2,3,4,5]
sales_2024 = [100, 150, 120, 180, 200]
sales_2025 = [130, 140, 160, 170, 220]

plt.plot(days, sales_2024, marker='o', label='2024', color='yellow')
plt.plot(days, sales_2025, marker='o', label='2025', color='red')
plt.title("Sales Comparison: 2024 vs 2025")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
# example
days = [1,2,3,4,5]
temps_city_A = [20, 22, 21, 23, 25]
temps_city_B = [15, 17, 16, 18, 20]

plt.plot(days, temps_city_A, marker='o', label='City A', color='yellow')
plt.plot(days, temps_city_B, marker='o', label='City B', color='red')
plt.title("Temperature Comparison: City A vs City B")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.legend()
plt.show()
```
