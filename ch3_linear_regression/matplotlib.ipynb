```python
import matplotlib.pyplot as plt
```

```python
# NOTES
# Matplotlib is a Python library for drawing charts like line graphs, bar chats, scatter plots
# ...etc. It's the paper and pencil of the python world. Pandas gets your data organized,
# while matplotlib is what you actually use to draw a picture of it.

# Matplotlib gives you a canvas (called a figure),
# a drawing tool (the plt module)
# and functions to draw stuff on the canvas/figure.

# nothing appears on the screen until you tell it to, i.e. you build the picture step by step
# then you say "show me the result".
```

```python
# the 2 functions you need to know for the plt module:

# plt.plot() builds the graph in memory - it doesn't display it yet

# plt.show() actually displays the image/graph.
```

```python
# example
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

# build the graph in memory.
plt.plot(x, y)

# display the graph.
plt.show()
```

```python
# example 2
# A small shop's sales over 5 days:
# Days 1,2,3,4,5
# Sales 100, 150, 120, 180, 200

days = [1, 2, 3, 4, 5]
sales = [100, 150, 120, 180, 200]

# define the axes labels.
plt.xlabel("Days")
plt.ylabel("Sales")

# define the title of the graph.
plt.title("Shop Sales This Week")

# Display a grid.
plt.grid(True)

# build the graph in memory.
plt.plot(days, sales)

# display the graph.
plt.show()
```

```python
# Controlling the axis range.
# By default, matplotlib zooms the chart to fit your data, i.e. the x-axis starts
# at the smallest x-value and ends at the largest, same for the y-axis.
# sometimes you want control over that yourself.

# you use the plt.xlim() and plt.ylim() functions, respectively.
plt.xlim(xmin=0, xmax=6)
plt.ylim(ymin=0, ymax=300)

# realised the labels, grid ant titles are missing.
plt.title("Sales This Week")

# Define labels for the graph.
plt.xlabel("Days")
plt.ylabel("Sales")

# Add a grid
plt.grid(True)

# Build the graph in memory.
plt.plot(days, sales)

# Display the graph.
plt.show()
```

```python
# Custom tick labels.
# The labels for the days, 1,2,3,4,5.. .etc are not as meaninful as "Monday", "Tuesday"...etc
# To adjust the tick labels you use the plt.xticks() and plt.yticks() functions.
# sales = [100, 150, 120, 180, 200]

# Define the tick labels.
day_of_the_week = ["Mon", "Tue", "Wed", "Thu", "Fri"]
monetary_values = ["$0", "$50", "$100", "$150", "$200"]
plt.xticks(days, day_of_the_week)
plt.yticks([0, 50, 100, 150, 200], monetary_values)

# Add a title.
plt.title("Sales This Week")

# Define the axes limits.
plt.xlim(xmin=0, xmax=6)
plt.ylim(ymin=0, ymax=200)
# Add axes labels.
plt.xlabel("Days")
plt.ylabel("Sales")

# Add a grid
plt.grid(True)

# Build the graph in memory.
plt.plot(days, sales)

# Display the graph.
plt.show()
```

```python
# Markers
# A marker puts a visible symbol where each data point is.
# to add a marker, pass a third argument to plot which represents one of the following markers:

# - 'o', 's', '^', '*', 'x'

# NOTE: if you use a keyword argument for the marker, plt keeps the line and adds a dot on top
# of it. However, if you use a positional argument, plt drops the line and just plots markers.

plt.title("Shop Sales This Week")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.grid(True)
plt.plot(days, sales, marker="o")
plt.show()
```

```python
# NOTE: plt.show() clears/resets the canvas so you can start a fresh empty chart.
# The takeaway is to always put a chart's entire set of instructions in one cell.
```
